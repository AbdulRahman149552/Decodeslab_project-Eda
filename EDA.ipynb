```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 5))

plt.plot(monthly_revenue.index.astype(str), monthly_revenue.values)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
monthly_orders = (
    clean_df.groupby("YearMonth")["OrderID"]
    .nunique()
)

monthly_orders
```

```python
plt.figure(figsize=(12, 5))

plt.plot(monthly_orders.index.astype(str), monthly_orders.values)

plt.title("Monthly Order Volume")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
product_revenue = product_revenue.sort_values()

plt.figure(figsize=(8, 5))

plt.barh(product_revenue.index, product_revenue.values)

plt.title("Revenue by Product")
plt.xlabel("Revenue")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```
